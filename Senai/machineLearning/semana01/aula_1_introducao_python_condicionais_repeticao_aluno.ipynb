## 8. Exercício com temperatura de máquina

Peça a temperatura e classifique:
- acima de 90 → alerta crítico
- entre 75 e 89.9 → atenção
- abaixo de 75 → normal


```python
# TODO: pedir a temperatura com input
temperatura=input('Informe a temperatura')
# TODO: converter para float
temperatura=float(temperatura)
# TODO: usar if / elif / else
if temperatura>90:
  print('Alerta crítico')
elif temperatura>=75 and temperatura<=89.9:
  print('Atenção')
else:
  print('Normal')
```

## 13. Exemplo com temperaturas industriais

Use a lista abaixo e classifique cada temperatura:
- 90 ou mais → alerta crítico
- 75 até 89 → atenção
- abaixo de 75 → normal


```python
temperaturas = [68, 72, 91, 77, 95]
for temperatura in temperaturas:
# TODO: usar for + if / elif / else
  if temperatura>=90:
    print('Alerta crítico')
  elif temperatura>=70 and temperatura<90:
    print('Atenção')
  else:
    print('Normal')
```

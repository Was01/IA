## 7. Função com coleção

```python
# TODO: criar função que recebe lista e percorre com for
def percorreLista(lista):
  for item in lista:
# TODO: aplicar if para classificar valores
    if item<20:
      print('Baixa')
    elif item<30:
      print('Média')
    else:
      print('Alta')

lista=[50,30,10]
percorreLista(lista)
```

## 8. Desafio Final

```python
# TODO: criar função própria que:
# 1. receba uma coleção
# 2. percorra os dados
# 3. faça uma análise simples
lista=[2,3,5,8,10]

def primo(n):
    if n == 1:
        return False
    else:
        for i in range(2, n):
            if n % i == 0:
                return False
    return True

```

```python
# Espaço livre para testes
[primo(n) for n in lista]
```

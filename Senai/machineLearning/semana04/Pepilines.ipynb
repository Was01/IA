```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler


import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

```python
caminho='/content/german_credit_data.csv'

df=pd.read_csv(caminho)
```

```python
df.head()
```

```python
df['Risk']=np.where(df['Risk']=='bad',1,0)
df.head()
```

```python
novo_df=df[['Age','Credit amount','Duration','Risk']].copy()
novo_df

```

```python
fetures=novo_df.drop('Risk',axis=1)
labels=df['Risk']
```

```python
X_train,X_test,y_train,y_test=train_test_split(fetures,labels,test_size=0.3,random_state=0)
```

```python
pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
    ('clf', DecisionTreeClassifier(random_state=42,max_depth=4))
])
pipe
```

```python
pipe.fit(X_train,y_train)
y_predict=pipe.predict(X_test)
```

```python
# 14. Métricas de Avaliação
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
acc = accuracy_score(y_test, y_predict)

print("Acurácia na validação:", round(acc, 4))
print("\nRelatório de classificação:")
print(classification_report(y_test,y_predict ))
```

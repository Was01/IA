```python
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

df = pd.DataFrame({
    "idade":   [25, 32, np.nan, 41],
    "salario": [3000, np.nan, 5000, 8000],
    "cidade":  ["SP", "RJ", np.nan, "SP"],
    "cargo":   ["analista", "gerente", "analista", np.nan]
})


```

```python
df.head()
```

```python
colunas_numericas=df.select_dtypes(include='number').columns.tolist()
colunas_categoricas=df.select_dtypes(include='object').columns.tolist()
```

```python
print(colunas_numericas)
print(colunas_categoricas)
```

```python
pipeline_numerica=Pipeline([('imputer',SimpleImputer(strategy='median')),('scaler',StandardScaler())])
```

```python
pipeline_categorica = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessador = ColumnTransformer([
    ("num", pipeline_numerica, colunas_numericas),
    ("cat", pipeline_categorica, colunas_categoricas)
])
```

```python
dados_transformados = preprocessador.fit_transform(df)
```

```python
df_final = pd.DataFrame(
    dados_transformados,
    columns=preprocessador.get_feature_names_out(),
    index=df.index
)
```

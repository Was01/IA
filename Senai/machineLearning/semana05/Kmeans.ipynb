```python
# Importando o dataset Iris

from sklearn.datasets import load_iris

import pandas as pd
```

```python
# Carregando o dataset iris

iris=load_iris()

df=pd.DataFrame(data=iris.data,columns=iris.feature_names)

df.head()
```

```python
# Selecionando apenas as colunas de pétalas

X=df[['petal length (cm)','petal width (cm)']]

X.head()
```

```python
# Visualizando graficamente os pontos

import matplotlib.pyplot as plt

fig, ax=plt.subplots()

ax.scatter(X['petal length (cm)'],X['petal width (cm)'])

ax.set_xlabel('Comprimento da pétala (cm)')
ax.set_ylabel('Largura da pétala (cm)')


plt.show()
```

```python
# Importando o KMeans

from sklearn.cluster import KMeans
```

```python
# Utilizando o algoritmo

kmeans2=KMeans(n_clusters=2,random_state=0).fit(X)
```

```python
# Verificando quais foram os labels determinados pelo k-means

kmeans2.labels_
```

```python
fig, ax=plt.subplots()

ax.scatter(X['petal length (cm)'],X['petal width (cm)'], c=kmeans2.labels_)

ax.set_xlabel('Comprimento da pétala (cm)')
ax.set_ylabel('Largura da pétala (cm)')
```

```python
# Utilizando k=3

kmeans3=KMeans(n_clusters=3,random_state=0).fit(X)
```

```python
# Método do cotovelo

valores_k=[]
inercias=[]

# Percorrendo pra diferentes valores de k

for i in range(1,15):
  kmeans=KMeans(n_clusters=i,random_state=0).fit(X)
  valores_k.append(i)
  inercias.append(kmeans.inertia_)
```

```python
fig, ax=plt.subplots()

ax.plot(valores_k,inercias)

ax.set_xlabel('Valores_k')
ax.set_ylabel('Inercia')
```

```python
# Coeficiente silhueta

from sklearn.metrics import silhouette_score

silhouette_score(X,kmeans2.labels_)

```

```python

from sklearn.metrics import silhouette_score

silhouette_score(X,kmeans3.labels_)
```

```python
# Percorrendo para diferentes valores de k

valores_k=[]
s=[]

for i in range(2,15):
  kmeans=KMeans(n_clusters=i,random_state=0).fit(X)
  valores_k.append(i)
  s.append(silhouette_score(X,kmeans.labels_))
```

```python
fig, ax=plt.subplots()

ax.plot(valores_k,s)

ax.set_xlabel('Valores_k')
ax.set_ylabel('Coeficiente Silhouette')
```

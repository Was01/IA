```python
import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer
```

```python
caminho='/content/svbr.csv'
```

```python
df=pd.read_csv(caminho,sep=';')

df.head()
```

```python
df.isna().sum()
```

```python


X = df.drop(columns=['Canal'])          # só Inscritos e Visualizações
```

```python


imputer=SimpleImputer(missing_values=np.nan,strategy='mean')

X = pd.DataFrame(imputer.fit_transform(X), columns=X.columns, index=X.index)

X.insert(0, 'Canal',df['Canal'].values)
print(X.values)
```

```python
print(X.isna().sum())   # deve ser tudo 0 nas colunas numéricas
print(X.dtypes)         # Canal = object, as demais = float64
```

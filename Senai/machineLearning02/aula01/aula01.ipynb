```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("ColabPySpark")
    .master("local[*]")
    .getOrCreate()
)

# SparkContext é a porta de entrada para a API, principalmente RDD
sc = spark.sparkContext
```

```python
# Não printa coisas demais

sc.setLogLevel('WARN')
```

```python
# Raiz quadrada de números de uma lista

numeros=sc.parallelize([1,2,3,4])

# map é uma transformação

raiz_quadrada=numeros.map(lambda x: x*x)
```

```python
raiz_quadrada.collect()
```

```python
rdd =sc.parallelize([1,2,3])

print(rdd.count())
print(rdd.countByValue())
print(rdd.take(2))
```

## Histograma do MovieLens

```python
path='/content/u.data'
```

```python
!ls
```

```python
# Criar um RDD

lines=sc.textFile(path)
```

```python
# Quebrar linhas por TAB
# parts é um RDD

parts=lines.map(lambda line: line.split('\t'))
```

```python
# filtrar linhas

parts_ok=parts.filter(lambda p:len(p)==4)
```

```python
# Obtem apenas o rating

ratings=parts_ok.map(lambda p:int(p[2]))
```

```python
# Criando pares chave e valor: (rating,1)

ratings_ones=ratings.map(lambda r: (r,1))
```

```python
# Somar por chave (agregação distribuída)

counts_rdd=ratings_ones.reduceByKey(lambda a,b:a+b)
```

```python
counts_list=counts_rdd.collect()
```

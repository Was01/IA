```python
# Importa as bibliotecas pandas (manipulação de dados) e NumPy (cálculos numéricos)

import pandas as pd
import numpy as np
```

```python
# Define o caminho do arquivo CSV com os dados de admissão

caminho='/content/admission.csv'
```

```python
# Lê o CSV (separado por ponto e vírgula) em um DataFrame e exibe as 5 primeiras linhas

baseDeDados=pd.read_csv(caminho,delimiter=';')

baseDeDados.head()
```

```python
# Separa as variáveis de entrada (todas as colunas, exceto a última) em um array NumPy e exibe


X=baseDeDados.iloc[:,:-1].values
print(X)
```

```python
# Separa a variável alvo (última coluna) em um array NumPy e exibe

y=baseDeDados.iloc[:,-1].values
print(y)
```

```python
# Preenche valores ausentes (NaN) das colunas de X, a partir da segunda, com a mediana de cada coluna e exibe


from sklearn.impute import SimpleImputer

imputer = SimpleImputer(missing_values=np.nan,strategy='median')

X[:,1:]=imputer.fit_transform(X[:,1:])

print(X)
```

```python
# Converte os valores de texto da primeira coluna de X em números inteiros (codificação de categorias) e exibe


from sklearn.preprocessing import LabelEncoder

labelencoder_X = LabelEncoder()
X[:,0] = labelencoder_X.fit_transform(X[:,0])

print(X)
```

```python
# Cria colunas dummy (one-hot encoding) a partir da primeira coluna de X, com 0 e 1 para cada categoria, e exibe

D=pd.get_dummies(X[:,0],dtype=int)
print(D.values)
```

```python
# Substitui a primeira coluna de X pelas colunas dummy (D), unindo-as às demais colunas, e exibe



X=np.concatenate((D.values,X[:,1:]),axis=1)

print(X)
```

```python
# Divide os dados em treino (80%) e teste (20%) para as variáveis de entrada (X) e o alvo (y)

from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test=train_test_split(X,y,test_size=0.2)
```

```python
# Padroniza as variáveis (média 0 e desvio padrão 1) e aplica a escala ao treino e ao teste

from sklearn.preprocessing import StandardScaler

scaleX= StandardScaler()

X_train= scaleX.fit_transform(X_train)

X_teste=scaleX.fit_transform(X_train)
```

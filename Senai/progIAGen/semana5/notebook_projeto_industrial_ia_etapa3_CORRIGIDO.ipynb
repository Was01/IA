## Importando Bibliotecas

```
import pandas as pd
import google.generativeai as genai # caso encontre algum erro ao iniciar essa lib, siga as ações das aulas anteriores.
```

## Configuração da API KEY

Crie sua chave no Google AI Studio:
https://aistudio.google.com/

```
# Configuração da API KEY
# RECOMENDADO: nunca deixe a chave escrita direto no código.
#
# No Google Colab: Menu lateral (ícone de chave) -> Secrets -> adicione GOOGLE_API_KEY
# from google.colab import userdata
# GOOGLE_API_KEY = userdata.get("GOOGLE_API_KEY")
#
# No VSCode/local: crie um arquivo .env com GOOGLE_API_KEY=sua_chave_aqui
# e use python-dotenv para carregar:
import os
from dotenv import load_dotenv

load_dotenv()
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")

if not GOOGLE_API_KEY:
    raise ValueError(
        "GOOGLE_API_KEY não encontrada. Configure a variável de ambiente "
        "ou o Secret do Colab antes de continuar."
    )

genai.configure(api_key=GOOGLE_API_KEY)
```

## Inicializando o Modelo Gemini

```
# Defina qual modelo você vai usar
model = genai.GenerativeModel("gemini-2.0-flash")
```

## Primeiro Teste com IA

```
# teste a api
response = model.generate_content(
    "Explique rapidamente o que é manutenção preditiva industrial."
)

print(response.text)
```

## Selecionando um Registro

```
registro = df.iloc[10]

print(registro)
```

## Criando um Prompt Inteligente

```
prompt = f'''
Você é um especialista em manutenção industrial.

Analise os dados abaixo:

Temperatura: {registro['temperatura']}
Vibração: {registro['vibracao']}
Pressão: {registro['pressao']}
Tempo de operação: {registro['tempo_operacao']}

Explique:
1. Se existe risco de falha;
2. Qual sensor aparenta maior criticidade;
3. Sugestões de manutenção preventiva.
'''
```

## Enviando Dados ao Gemini

```
response = model.generate_content(prompt)

print(response.text)
```

## Integração com Machine Learning

```
# Exemplo fictício (Integre com o seu código das etapas anteriores)
previsao = 1

prompt_ml = f'''
O sistema de Machine Learning detectou:

Previsão de falha: {previsao}

Explique:
- O que essa previsão significa;
- Quais ações devem ser tomadas;
- Possíveis riscos industriais.
'''

response = model.generate_content(prompt_ml)

print(response.text)
```

## Geração Automática de Relatórios

```
prompt_relatorio = f'''
Crie um relatório técnico resumido com base nos dados:

Temperatura: {registro['temperatura']}
Vibração: {registro['vibracao']}
Pressão: {registro['pressao']}
Tempo de operação: {registro['tempo_operacao']}

O relatório deve conter:
- Diagnóstico;
- Grau de risco;
- Recomendação técnica.
'''

response = model.generate_content(prompt_relatorio)

print(response.text)
```

## Analisando Múltiplos Registros

```
for i in range(3):
    registro = df.iloc[i]

    prompt = f'''
    Analise os sensores industriais abaixo:

    Temperatura: {registro['temperatura']}
    Vibração: {registro['vibracao']}
    Pressão: {registro['pressao']}
    Tempo de operação: {registro['tempo_operacao']}

    Gere um diagnóstico técnico resumido.
    '''

    response = model.generate_content(prompt)

    print(f"===== REGISTRO {i} =====")
    print(response.text)
    print()
```

## Salvando Relatórios

```
with open("relatorio_diagnosticos.txt", "w", encoding="utf-8") as arquivo:

    for i in range(5):
        registro = df.iloc[i]

        prompt = f'''
        Analise os sensores:

        Temperatura: {registro['temperatura']}
        Vibração: {registro['vibracao']}
        Pressão: {registro['pressao']}
        Tempo de operação: {registro['tempo_operacao']}

        Gere um diagnóstico técnico.
        '''

        response = model.generate_content(prompt)

        arquivo.write(f"===== REGISTRO {i} =====\n")
        arquivo.write(response.text)
        arquivo.write("\n\n")

print("Relatório salvo com sucesso!")
```

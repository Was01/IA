```python
from PIL import Image,ImageDraw
import os
```

```python
def triangle(size):
  white=(255,255,255)
  black=(0,0,0)

  img=Image.new('RGB',(size,size), white)

  for x in range(size):
    for y in range(size):
      if x<y:
        img.putpixel((x,y),black)
  return img
```

```python
t=triangle(400)
display(t)
```

```python
def bandeira_franca(height):
  ratio=3/2
  width=int(height*ratio)

  blue=(0,85,164)
  white=(255,255,255)
  red=(239,65,53)

  image=Image.new('RGB',(width,height),white)

  offset=width//3

  for x in range(offset):
    for y in range(height):
      image.putpixel((x,y),blue)

  for x in range(2*offset,width):
    for y in range(height):
      image.putpixel((x,y),red)

  return image






```

```python
franca=bandeira_franca(400)
display(franca)
```

```python
def bandeira_japao(height):
  width=3*height//2

  white=(255,255,255)
  red=(173,35,51)
  black=(0,0,0)


  img=Image.new('RGB',(width,height),white)

  # Objeto para desenhar sobre a imagem
  draw = ImageDraw.Draw(img)

  # Desenha um retângulo: coordenadas (x0, y0, x1, y1)
  draw.rectangle([0, 0, width, height], fill='white', outline='black', width=2)




  r=3*height//10
  c=(width//2,height//2)


  for x in range(c[0]-r,c[0]+r):
    for y in range(c[1]-r,c[1]+r):
      if (x-c[0])**2+(y-c[1])**2<=r**2:
        img.putpixel((x,y),red)

  return img






```

```python
japao=bandeira_japao(400)
display(japao)
```

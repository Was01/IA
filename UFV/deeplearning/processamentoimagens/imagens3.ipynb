## Convertendo uma imagem para escala de cinza

```python
from PIL import Image
```

```python

image=Image.open('/content/richardmiddelbos-butterfly-4873368.jpg')
```

```python
img = image.resize((600, 600))
display(img)
```

```python
print(img.getpixel((250,50)))
```

```python
newimage=Image.new('RGB',(50,50),(97,146,46))
display(newimage)
```

```python
def grayscale(imagem,width,height):
  for x in range(int(width)):
    for y in range(int(height)):
      pixel=imagem.getpixel((x,y))
      lum=(pixel[0]+pixel[1]+pixel[2])//3
      imagem.putpixel((x,y),(lum,lum,lum))
  return imagem
```

```python
t=grayscale(img,600,600)
display(t)
```

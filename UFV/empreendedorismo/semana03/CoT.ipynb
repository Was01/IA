```python
from google import genai
from google.colab import userdata

api_key = userdata.get('Gemini_API_Key')
client = genai.Client(api_key=api_key)
```

```python
# Função auxiliar para fazer chamadas à API do Gemini
def obter_resposta_llm(texto_prompt, modelo="gemini-3.6-flash"):
    if not client:
        return "Erro: O cliente da API não foi inicializado."

    try:
        resposta = client.models.generate_content(
            model=modelo,
            contents=texto_prompt,
        )
        return resposta.text
    except Exception as e:
        return f"Erro na comunicação com a API do Gemini: {e}"
```

Problema:

Resolva o problema abaixo, mostrando cada etapa do raciocínio antes

Problema: Um carro faz metade de um percurso a 60 km/h e a outra
metade (mesma distância) a 90 km/h. Qual é a velocidade média do
carro no percurso todo?

Siga estas etapas:
1. Defina uma distância total arbitrária (escolha uma incógnita)
2. Calcule o tempo gasto em cada metade do percurso
3. Calcule o tempo total e a distância total
4. Calcule a velocidade média (distância total / tempo total)

Resposta do LLM (Prompt Padrao):
Aqui está a resolução do problema passo a passo, conforme solicitado:

---

### **Etapa 1: Definição da distância**

Vamos definir uma incógnita para a distância total do percurso:
* Vamos chamar a **distância total** de $D$.
* Como o percurso é dividido em duas partes iguais, cada metade terá uma distância de $\frac{D}{2}$.

---

### **Etapa 2: Cálculo do tempo gasto em cada metade do percurso**

A relação entre tempo ($t$), distância ($d$) e velocidade ($v$) é dada por $t = \frac{d}{v}$.

* **Primeira metade do percurso:**
  * Distância: $\frac{D}{2}$
  * Velocidade: $60 \text{ km/h}$
  * Tempo $t_1 = \frac{\frac{D}{2}}{60} = \frac{D}{120} \text{ horas}$

* **Segunda metade do percurso:**
  * Distância: $\frac{D}{2}$
  * Velocidade: $90 \text{ km/h}$
  * Tempo $t_2 = \frac{\frac{D}{2}}{90} = \frac{D}{180} \text{ horas}$

---

### **Etapa 3: Cálculo do tempo total e da distância total**

* **Distância total ($D_{\text{total}}$):**
  $$D_{\text{total}} = D$$

* **Tempo total ($T_{\text{total}}$):**
  $$T_{\text{total}} = t_1 + t_2 = \frac{D}{120} + \frac{D}{180}$$

  Para somar as frações, encontramos o Mínimo Múltiplo Comum (MMC) entre $120$ e $180$, que é $360$:
  $$T_{\text{total}} = \frac{3D}{360} + \frac{2D}{360} = \frac{5D}{360}$$

  Simplificando a fração por $5$:
  $$T_{\text{total}} = \frac{D}{72} \text{ horas}$$

---

### **Etapa 4: Cálculo da velocidade média**

A velocidade média ($V_m$) é dada pela razão entre a distância total e o tempo total:

$$V_m = \frac{D_{\text{total}}}{T_{\text{total}}} = \frac{D}{\frac{D}{72}}$$

Multiplicando pelo inverso da fração do denominador:

$$V_m = D \cdot \frac{72}{D} = 72 \text{ km/h}$$

---

### **Etapa 5: Comparação com a média aritmética e explicação**

* **Média aritmética simples:**
  $$\frac{60 + 90}{2} = \frac{150}{2} = 75 \text{ km/h}$$

* **Comparação:**
  A velocidade média real ($72 \text{ km/h}$) é **menor** que a média aritmética simples ($75 \text{ km/h}$).

* **Explicação:**
  A velocidade média é uma média ponderada em relação ao **tempo**, não em relação à distância. Como a primeira metade do percurso foi feita a uma velocidade menor ($60 \text{ km/h}$), o carro levou **mais tempo** para percorrer essa metade do que para percorrer a segunda metade ($90 \text{ km/h}$).
  Como o carro passou mais tempo dirigindo a $60 \text{ km/h}$, essa velocidade menor "puxa" a média final para mais perto dela.

*(Obs: Quando as distâncias são iguais, a velocidade média correta é dada pela **média harmônica** das velocidades).*

---

### **Resposta Final**
A velocidade média do carro no percurso todo é de **$72 \text{ km/h}$**.

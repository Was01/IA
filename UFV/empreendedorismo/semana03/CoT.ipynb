```python
from google import genai
from google.colab import userdata

api_key = userdata.get('Gemini_API_Key')
client = genai.Client(api_key=api_key)
```

```python
# Função auxiliar para fazer chamadas à API do Gemini
def obter_resposta_llm(texto_prompt, modelo="gemini-3.6-flash"):
    if not client:
        return "Erro: O cliente da API não foi inicializado."

    try:
        resposta = client.models.generate_content(
            model=modelo,
            contents=texto_prompt,
        )
        return resposta.text
    except Exception as e:
        return f"Erro na comunicação com a API do Gemini: {e}"
```

```python
problema_logico='''
Problema: Resolver 3x + 7 = 22

Vamos resolver passo a passo:

Objetivo: isolar o x sozinho de um lado da equação.

Passo1: eliminar o +7: Como 7 está sendo somado do lado esquerdo, faço a operação inversa (subtração) dos dois lados:
Passo2: eliminar o 3 que multiplica x: Como 3 está multiplicando, faço a operação inversa (divisão) dos dois lados:
Paso3 Verificação: Verificação: substitua o valor de x encontrado  na equação original:
'''
```

```python

if client:
    resposta_padrao = obter_resposta_llm(problema_logico)
    print("Problema:\n" + problema_logico)
    print("\nResposta do LLM (Prompt Padrao):\n" + resposta_padrao)
```

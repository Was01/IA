```python
from google import genai
from google.colab import userdata

api_key = userdata.get('Gemini_API_Key')
client = genai.Client(api_key=api_key)
```

```python
# Função auxiliar para fazer chamadas à API do Gemini
def obter_resposta_llm(texto_prompt, modelo="gemini-3.6-flash"):
    if not client:
        return "Erro: O cliente da API não foi inicializado."

    try:
        resposta = client.models.generate_content(
            model=modelo,
            contents=texto_prompt,
        )
        return resposta.text
    except Exception as e:
        return f"Erro na comunicação com a API do Gemini: {e}"
```

```python
problema_logico='''
O Departamento de Biologia da universidade possui um orçamento total de
R$ 15.000,00 para a compra de novos equipamentos. Dentro desse valor,
eles receberam uma verba específica de R$ 4.000,00 que só pode ser
usada para "equipamentos de laboratório genômico". Eles precisam decidir
quais dos seguintes itens podem comprar:
●​ Microscópio Confocal: R$ 8.000,00
●​ Centrífuga Refrigerada: R$ 5.000,00
●​ Sequenciador de DNA Portátil: R$ 3.500,00 (considerado
equipamento de laboratório genômico)
●​ Sistema de Eletroforese: R$ 2.000,00

Pergunta: Quais equipamentos o departamento pode comprar, garantindo
que a verba de genômica seja usada para o Sequenciador de DNA e que o
orçamento total não seja excedido?
'''
```

```python
print("\n--- Tentativa 1: Resolvendo com Prompt Padrao ---")
if client:
    resposta_padrao = obter_resposta_llm(problema_logico)
    print("Problema:\n" + problema_logico)
    print("\nResposta do LLM (Prompt Padrao):\n" + resposta_padrao)
```

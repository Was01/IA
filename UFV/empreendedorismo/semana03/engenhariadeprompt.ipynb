```python
from google import genai
from google.colab import userdata

api_key = userdata.get('Gemini_API_Key')
client = genai.Client(api_key=api_key)

```

```python
# Função auxiliar para fazer chamadas à API do Gemini
def obter_resposta_llm(texto_prompt, modelo="gemini-3.6-flash"):
    if not client:
        return "Erro: O cliente da API não foi inicializado."

    try:
        resposta = client.models.generate_content(
            model=modelo,
            contents=texto_prompt,
        )
        return resposta.text
    except Exception as e:
        return f"Erro na comunicação com a API do Gemini: {e}"
```

```python
trecho_de_noticias_universitarias = """
Título: Universidade Lança Nova Iniciativa de Sustentabilidade 'GreenFuture U'
Data: 10 de Setembro de 2024
Pesquisadores da Escola de Ciências Ambientais e do Departamento de
Engenharia anunciaram hoje uma nova e importante iniciativa de
sustentabilidade em todo o campus. A iniciativa 'GreenFuture U' visa
fazer a transição da universidade para fontes de energia 100%
renováveis até 2035, incluindo planos para instalações de painéis
solares e projetos de energia geotérmica.
"""
```

```python
# Exemplo 1: Extraindo Departamentos (Código completo como exemplo)

prompt_extrair_departamentos = f"""
Extraia todos os departamentos ou escolas universitárias mencionados
no trecho de notícia abaixo.
Liste-os separados por vírgulas. Se não houver nenhum, escreva "Nenhum".

Notícia:
{trecho_de_noticias_universitarias}
"""
```

```python
if client:
    departamentos = obter_resposta_llm(prompt_extrair_departamentos)
    print(f"Departamentos/Escolas extraídos: {departamentos}")
```

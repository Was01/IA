```python
from google import genai
from google.colab import userdata

api_key = userdata.get('Gemini_API_Key')
client = genai.Client(api_key=api_key)
```

```python
# Função auxiliar para fazer chamadas à API do Gemini
def obter_resposta_llm(texto_prompt, modelo="gemini-3.6-flash"):
    if not client:
        return "Erro: O cliente da API não foi inicializado."

    try:
        resposta = client.models.generate_content(
            model=modelo,
            contents=texto_prompt,
        )
        return resposta.text
    except Exception as e:
        return f"Erro na comunicação com a API do Gemini: {e}"
```

```python
texto_prompt='''
Exemplo 1:
Tópico: Área de triângulos
Problema: Um terreno triangular tem base 12m e altura 8m. Qual sua área?
Resposta: A = (base × altura) / 2 = (12 × 8) / 2 = 48 m²

Exemplo 2:
Tópico: Perímetro de retângulos
Problema: Um jardim retangular mede 5m de largura e 9m de comprimento. Qual seu perímetro?
Resposta: P = 2 × (largura + comprimento) = 2 × (5 + 9) = 28 m

Agora crie um problema no mesmo estilo:
Tópico: Volume de cubos
'''
```

```python
if client:
    solucao = obter_resposta_llm(texto_prompt)
    print(f"Problema/Solução: {solucao}")
```

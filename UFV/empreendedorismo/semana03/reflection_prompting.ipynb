```python
from google import genai
from google.colab import userdata

api_key = userdata.get('Gemini_API_Key')
client = genai.Client(api_key=api_key)
```

```python
# Função auxiliar para fazer chamadas à API do Gemini
def obter_resposta_llm(texto_prompt, modelo="gemini-3.6-flash"):
    if not client:
        return "Erro: O cliente da API não foi inicializado."

    try:
        resposta = client.models.generate_content(
            model=modelo,
            contents=texto_prompt,
        )
        return resposta.text
    except Exception as e:
        return f"Erro na comunicação com a API do Gemini: {e}"
```

```python
rascunho_inicial='''Participe do nosso hackathon universitário, será no
 próximo mês. Haverá codificação. '''

print("\n---Processo de Reflexão---")
print(f'Rascunho Inicial: {rascunho_inicial}')
```

```python
prompt_de_critica=f'''Critique o anúncio {rascunho_inicial} levando em
consideração os seguintes pontos:
É envolvente e engajador para os alunos?
Declara claramente o prósito do evento?
Especifica quais atividades serão oferecidas?
O tom é apropriado para incentivar a participação dos alunos?
'''
```

```python
if client:
  critica=obter_resposta_llm(prompt_de_critica)
  print(f'Critíca: {critica}')
```

```python
prompt_final_refinado=f'''Com base no {rascunho_inicial} e na {critica}, reecreva completamente o anúncio do hackaton,
O novo anúncio deve ser energético, inspirador e informativo.
'''
```

```python
if client:
  anuncio_final=obter_resposta_llm(prompt_final_refinado)
  print(f'Anúncio Final: {anuncio_final}')
```

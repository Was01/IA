```python
import pandas as pd
df=pd.read_csv('homeprices.csv')
print(df)
```

```python
df.plot(x='area',y='price',kind='scatter');
```

```python
from sklearn import linear_model

model=linear_model.LinearRegression();
model.fit(df[['area']],df[['price']]);
```

```python
model.predict(pd.DataFrame({'area':[250]}))
```

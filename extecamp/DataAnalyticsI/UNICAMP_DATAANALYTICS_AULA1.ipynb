Operadores de comparação em Python

```python
#exemplos de operadores de comparacao em Python
# o operador > (parentesis angular a direita)
#permite comparar 2 valores (ou variaveis) (ex: 5>10) para
#saber se o valor à esquerda é maior do que o da direita.
#o resultado é um valor boleano True ou False para a comparacao

print("5 é maior do que 10?", (5>10))

x=(5<10)
print("5 é menor do que 10?", x)

print("5 é igual a 10?", (5==10))

a=5  #dado pode vir de uma coluna de uma tabela
b=10 #dado pode vir de uma coluna de uma tabela
print("Operacao de igualdade", (a==b)) #operador de igualdade ==
print("Operador de maior", (a>b))  #operador de maior >
print("5 é maior ou igual a 10?", (a>=b))  #operador de maior ou igual >=
print("5 é menor a 10?", (a<b))  #operador de menor <
print("5 é menor ou igual a 10?", (a<=b))  #operador de menor <=
print("Operador de diferenca", (a!=b))  #operador de diferenca !=

```

```python
#operadores logicos boleanos
x=(5>5)
y=(6!=4)

print("x=",x)
print("y=",y)

z=not x
print(" x E y sao verdadeiros?", (x and y))
print(" x OU y sao verdadeiros?", (x or y))
print(" negar o valor de x:", z)
```

Estruturas condicionais: if, else, elif

```python
#estrutura de condicao if
#Python: é case-sensitive (maiusculas e minusculas sao tratadas de forma diferente)
x=4
print("Valor de x=",(x>5))
if x>5: #precisar ser True para executar o código da condição (x>5)
  print("x é maior do que 5:",x)

#Atividade: considere o valor de glicose que se altera entre pessoas.
#Faca um programa que, a partir do valor da glicose, imprima uma mensagem:
#glicose menor ou igual a 70: "Pessoa com hipoglicemia"
#glicose maior ou igual a 100: "Pessoa com hiperglicemia"
#glicose com outro valor: "Pessoa sem hipo ou hiperglicemia"
glicose=-10;

if glicose<=70:
  print("Pessoa com hipoglicemia: ", glicose)
elif glicose>=100:
  print("Pessoa com hiperglicemia: ", glicose)
else:
  print("Pessoa sem hipo ou hiperglicemia: ", glicose)


#faca um programa que diga se uma pessoa atingiu ou nao a maioridade:
idade=30
if idade>=18:
  print("Pessoa atingiu a maioridade")
else:
  print("Pessoa não atingiu a maioridade")
```

Estrutura de repetiçãoo for (laço de repetição)

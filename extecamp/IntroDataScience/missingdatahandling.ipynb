```python
import numpy as np
from sklearn.impute import KNNImputer

# Criando um conjunto de dados com valores ausentes
# Array de arrays: cada linha (array) é considerada uma amostra, logo o
# KNNImputer trabalha com essa premissa
data = np.array([[1,      2, np.nan],
                 [3,      4, 3],
                 [np.nan, 6, 5],
                 [8,      8, 7]])

# Inicializando o KNNImputer com 2 vizinhos
imputer = KNNImputer(n_neighbors=2)

# Aplicando o KNNImputer para imputar os valores ausentes
imputed_data = imputer.fit_transform(data)

print("Dados originais:\n", data)
print("Dados imputados:\n", imputed_data)
```

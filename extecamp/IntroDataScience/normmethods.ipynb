### Exemplos de Métodos de Normalização

1. **Min-Max Scaling**
   - Normaliza os dados para o intervalo [0, 1].
   - Fórmula: $$x' = \frac{x - \text{min}(X)}{\text{max}(X) - \text{min}(X)}$$
   - onde:
      - $x$: Valor original.
      - $x'$: Valor normalizado.
      - $\text{min}(X)$: Valor mínimo do atributo.
      - $\text{max}(X)$: Valor máximo do atributo.
   - Útil quando a distribuição dos dados é desconhecida e queremos uma escala fixa.

2. **Z-Score Normalization (Standardization)**
   - Normaliza os dados para ter média 0 e desvio padrão 1.
   - Fórmula: $$x' = \frac{x - \mu}{\sigma}$$
   - onde:
      - $x$: Valor original.
      - $x'$: Valor normalizado.
      - $\mu$: Média dos valores do atributo.
      - $\sigma$: Desvio padrão dos valores do atributo.
   - Útil para dados com distribuição normal (gaussiana).

3. **Max Abs Scaling**
   - Escala os dados pelo valor absoluto máximo do atributo, resultando em valores no intervalo [-1, 1].
   - Fórmula: $$x' = \frac{x}{\text{max}(|X|)}$$
   - onde:
      - $x$: Valor original.
      - $x'$: Valor normalizado.
      - $\text{max}(|X|)$: Valor absoluto máximo do atributo.
   - Útil para preservar a esparsidade dos dados.

4. **Robust Scaling**
   - Utiliza a mediana e o intervalo interquartil (IQR) para escalar os dados, sendo menos sensível a outliers.
   - Fórmula: $$x' = \frac{x - \text{mediana}(X)}{\text{IQR}(X)}$$
   - onde:
      - $x$: Valor original.
      - $x'$: Valor normalizado.
      - $\text{mediana}(X)$: Mediana dos valores do atributo.
      - $\text{IQR}(X)$: Intervalo interquartil dos valores do atributo.
   - Útil para dados com muitos outliers.

```python
# Importando as bibliotecas necessárias
from sklearn.preprocessing import MinMaxScaler, StandardScaler, MaxAbsScaler, RobustScaler

# Conjunto de dados exemplo
data = [[-1, 2], [-0.5, 6], [0, 10], [1, 18]]
print("Input Data:")
print(data)
```

```python
# 1. Min-Max Scaling
# Este método escala os dados para que todos os valores fiquem no intervalo [0, 1].
# É sensível a outliers, que podem distorcer a escala.
scaler = MinMaxScaler()
min_max_scaled_data = scaler.fit_transform(data)
print("Min-Max Scaling:")
print(min_max_scaled_data)
```

```python
# 2. Z-Score Normalization (Standardization)
# Este método transforma os dados para que tenham uma média de 0 e um desvio padrão de 1.
# É útil para dados com distribuição gaussiana (normal) e menos sensível a outliers.
scaler = StandardScaler()
standardized_data = scaler.fit_transform(data)
print("Z-Score Normalization (Standardization):")
print(standardized_data)
```

```python
# 3. Max Abs Scaling
# Este método escala os dados dividindo cada valor pelo valor absoluto máximo do atributo,
# resultando em valores no intervalo [-1, 1]. É útil para preservar a esparsidade dos dados.
scaler = MaxAbsScaler()
max_abs_scaled_data = scaler.fit_transform(data)
print("Max Abs Scaling:")
print(max_abs_scaled_data)
```

```python
# 4. Robust Scaling
# Este método utiliza a mediana e o intervalo interquartil para escalar os dados, sendo menos sensível a outliers.
# É útil para dados com muitos outliers.
scaler = RobustScaler()
robust_scaled_data = scaler.fit_transform(data)
print("Robust Scaling:")
print(robust_scaled_data)
```

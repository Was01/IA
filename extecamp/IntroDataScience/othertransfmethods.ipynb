### Agregação

#### Descrição

A **agregação** de dados é o processo de combinar dois ou mais atributos em um único atributo. Isso pode ser feito somando, tirando a média, calculando a mediana, ou utilizando outras funções de agregação. A agregação é útil para reduzir a dimensionalidade dos dados, o que pode simplificar a análise e melhorar o desempenho dos algoritmos de mineração de dados.

#### Objetivos

- **Reduzir a Dimensionalidade**: Diminuir o número de atributos, facilitando a visualização e a análise dos dados.
- **Simplificar o Modelo**: Tornar o modelo mais simples e interpretável, evitando overfitting.
- **Capturar Informações Relevantes**: Combinar informações de maneira significativa, preservando ou destacando padrões importantes.

#### Exemplo de Agregação

**Cenário**: Suponha que temos um conjunto de dados sobre transações de vendas com atributos como `price`, `quantity`, e `discount`. Podemos agregar esses atributos para criar um novo atributo `total_sales`, que representa o valor total da venda após aplicar o desconto.

**Antes da Agregação**:

| price | quantity | discount |
|-------|----------|----------|
| 100   | 2        | 0.1      |
| 150   | 1        | 0.05     |
| 200   | 3        | 0.2      |

**Depois da Agregação**:

| total_sales |
|-------------|
| 180         |
| 142.5       |
| 480         |

**Cálculo**:
$$ \text{total\_sales} = (\text{price} \times \text{quantity}) \times (1 - \text{discount}) $$


```python
import pandas as pd

# Dados de entrada
data = {
    'price': [100, 150, 200],
    'quantity': [2, 1, 3],
    'discount': [0.1, 0.05, 0.2]
}

df = pd.DataFrame(data)

# Agregar dados para calcular total_sales
df['total_sales'] = (df['price'] * df['quantity']) * (1 - df['discount'])

print(df)
```

### Geração de Atributos

#### Descrição

A **geração de atributos** (ou **feature engineering**) é o processo de criar novos atributos a partir dos atributos existentes nos dados brutos. Esses novos atributos são projetados para capturar melhor a informação presente nos dados, facilitando a descoberta de padrões e melhorando o desempenho dos algoritmos de mineração de dados.

#### Objetivos

- **Melhorar a Representatividade dos Dados**: Criar atributos que capturam informações relevantes de maneira mais eficaz.
- **Aumentar a Performance do Modelo**: Fornecer ao algoritmo de mineração de dados atributos mais informativos, resultando em melhores previsões e classificações.
- **Facilitar a Detecção de Padrões**: Ajudar a destacar relações e padrões nos dados que podem não ser imediatamente óbvios.

#### Exemplo de Geração de Atributos

**Cenário**: Suponha que temos um conjunto de dados com atributos como `height` (altura) e `weight` (peso). Podemos gerar um novo atributo chamado `bmi` (índice de massa corporal), que é uma medida mais relevante para várias análises de saúde.

**Antes da Geração de Atributos**:

| height | weight |
|--------|--------|
| 1.6    | 60     |
| 1.8    | 80     |

**Depois da Geração de Atributos**:

| height | weight | bmi       |
|--------|--------|-----------|
| 1.6    | 60     | 23.4375   |
| 1.8    | 80     | 24.6914   |

**Cálculo**:
$$ \text{bmi} = \frac{\text{weight}}{\text{height}^2} $$

```python
import pandas as pd

# Dados de entrada
data = {
    'height': [1.6, 1.8],
    'weight': [60, 80]
}

df = pd.DataFrame(data)

# Gerar novo atributo BMI
df['bmi'] = df['weight'] / (df['height'] ** 2)

print(df)
```

### Redução de Dimensionalidade

#### Descrição

A **redução de dimensionalidade** é o processo de reduzir o número de atributos (dimensões) em um conjunto de dados, mantendo a maior quantidade possível de informação relevante. Isso pode ser feito através de técnicas que transformam os dados em um espaço de menor dimensão ou selecionando um subconjunto dos atributos originais.

#### Objetivos

- **Simplificar o Modelo**: Tornar o modelo mais simples e interpretável, evitando overfitting.
- **Reduzir o Custo Computacional**: Diminuir o tempo de processamento e os recursos necessários para análise.
- **Evitar a Maldição da Dimensionalidade**: Melhorar a performance dos algoritmos de mineração de dados, que podem ser prejudicados por um grande número de dimensões.

#### Técnicas Comuns de Redução de Dimensionalidade

1. **Análise de Componentes Principais (PCA)**
   - **Descrição**: Transforma os dados para um novo espaço de menor dimensão, preservando a variância máxima.
   - **Objetivo**: Capturar a maior quantidade de variação nos dados com o menor número de componentes.

2. **Análise Discriminante Linear (LDA)**
   - **Descrição**: Encontra uma transformação linear que maximiza a separação entre diferentes classes.
   - **Objetivo**: Melhorar a discriminabilidade dos dados em tarefas de classificação.

3. **Seleção de Atributos**
   - **Descrição**: Seleciona um subconjunto dos atributos originais com base em sua relevância para a tarefa de mineração de dados.
   - **Objetivo**: Manter apenas os atributos mais informativos.

#### Exemplo de Redução de Dimensionalidade com PCA

**Cenário**: Suponha que temos um conjunto de dados com três atributos `A`, `B`, e `C`. Podemos aplicar PCA para reduzir esses

**Antes da Redução de Dimensionalidade**:

| A  | B  | C  |
|----|----|----|
| 2.5| 2.4| 0.5|
| 0.5| 0.7| 0.5|
| 2.2| 2.9| 0.5|
| 1.9| 2.2| 0.5|
| 3.1| 3.0| 0.5|

**Depois da Redução de Dimensionalidade**:

| PC1       | PC2       |
|-----------|-----------|
| 0.4436    | -0.2010   |
| -2.1772   | -0.0550   |
| 0.5707    | 0.3681    |
| -0.1290   | 0.0675    |
| 1.2919    | -0.1796   |

#### Significado dos Valores PC1 e PC ####
A ideia geral é que o PCA captura a covariância entre atributos (atributos dependentes), de forma que os dados mais importantes desses atributos possam ser representados por um único valor.
- **Exemplo:**
<figure>
   <image src="PCA_fish_shape.png" alt="Exemplo de uso do PCA" width="500">
   <figcaption style="font-size: 8px">Fonte: Werner and Friedrich 2014, fig. 1</figcaption>
</figure>

**PC1 (Primeiro Componente Principal)**

- **Descrição:** PC1 é a direção na qual os dados variam mais. É a linha que passa pelo centro dos dados e minimiza a distância ao quadrado dos pontos, captando a maior parte da variância presente nos dados originais

- **Importância:** Quanto maior o valor absoluto de PC1 para um ponto, mais esse ponto se afasta do centro da distribuição dos dados ao longo desta nova direção.

**PC2 (Segundo Componente Principal)**

- **Descrição:** PC2 é a direção ortogonal a PC1 que captura a maior quantidade de variância restante. É a segunda linha que passa pelo centro dos dados, captando a segunda maior parte da variância nos dados, independente da variância já capturada por PC1.

- **Importância:** O valor de PC2 representa a posição dos dados ao longo desta nova direção.


```python
import pandas as pd
from sklearn.decomposition import PCA

# Dados de entrada
data = {
    'A': [2.5, 0.5, 2.2, 1.9, 3.1],
    'B': [2.4, 0.7, 2.9, 2.2, 3.0],
    'C': [0.5, 0.5, 0.5, 0.5, 0.5]
}

df = pd.DataFrame(data)

# Aplicar PCA para reduzir para 2 componentes principais
pca = PCA(n_components=2)
principal_components = pca.fit_transform(df)

df_pca = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])

print(df_pca)
```

### Discretização

#### Descrição

A **discretização** é o processo de converter atributos contínuos em atributos discretos ou categóricos. Isso é feito dividindo o intervalo dos valores contínuos em um número finito de intervalos ou bins, e cada valor é substituído pelo identificador do intervalo correspondente.

#### Objetivos

- **Simplificação dos Dados**: Reduzir a complexidade dos dados contínuos, tornando-os mais manejáveis.
- **Facilitar a Interpretação**: Transformar dados contínuos em categorias pode facilitar a interpretação dos resultados.
- **Preparação para Modelos**: Alguns algoritmos de mineração de dados funcionam melhor com dados discretos.

#### Técnicas Comuns de Discretização

1. **Discretização por Intervalos Uniformes (Equal-width Binning)**
   - **Descrição**: Divide o intervalo de valores em `k` intervalos de largura igual.
   - **Objetivo**: Simples e fácil de implementar, mas pode não lidar bem com distribuições não uniformes dos dados.

2. **Discretização por Frequência Uniforme (Equal-frequency Binning)**
   - **Descrição**: Divide os dados em `k` intervalos, cada um contendo aproximadamente o mesmo número de observações.
   - **Objetivo**: Garante que cada intervalo tenha uma quantidade similar de dados, melhorando a representatividade.

3. **Discretização Baseada em Clusterização (Clustering-based Discretization)**
   - **Descrição**: Usa algoritmos de clusterização (como k-means) para agrupar dados contínuos em `k` clusters, e cada cluster é tratado como um intervalo.
   - **Objetivo**: Agrupa dados similares, preservando as características intrínsecas da distribuição.

#### Exemplo de Discretização por Intervalos Uniformes

**Cenário**: Suponha que temos um atributo contínuo `X` com os seguintes valores: `1.5`, `2.3`, `4.5`, `3.2`, `5.7`. Vamos discretizá-los em 3 intervalos uniformes.

**Antes da Discretização**:

| X  |
|----|
| 1.5|
| 2.3|
| 4.5|
| 3.2|
| 5.7|

**Após a Discretização (3 intervalos)**:

| X  | Bin  |
|----|------|
| 1.5| 1    |
| 2.3| 1    |
| 4.5| 3    |
| 3.2| 2    |
| 5.7| 3    |

```python
import numpy as np
import pandas as pd

# Dados de entrada
data = {'X': [1.5, 2.3, 4.5, 3.2, 5.7]}
df = pd.DataFrame(data)

# Discretização por intervalos uniformes
num_bins = 3
df['Bin'] = pd.cut(df['X'], bins=num_bins, labels=False) + 1  # Adiciona 1 para iniciar os bins em 1

print(df)
```

### Codificação de Categóricos

#### Descrição

A **codificação de categóricos** é o processo de transformar atributos categóricos em uma representação numérica que possa ser utilizada por algoritmos de mineração de dados e aprendizado de máquina. Isso é necessário porque muitos algoritmos requerem entradas numéricas.

#### Objetivos

- **Preparação dos Dados**: Converter dados categóricos para uma forma que possa ser usada por modelos de aprendizado de máquina.
- **Facilitar a Análise**: Tornar atributos categóricos utilizáveis em cálculos matemáticos e estatísticos.

#### Técnicas Comuns de Codificação de Categóricos

1. **Codificação de Rótulo (Label Encoding)**
   - **Descrição**: Cada valor categórico é convertido em um número inteiro único.
   - **Objetivo**: Simples e direta, mas pode introduzir ordens artificiais que não existem nos dados originais.

2. **Codificação One-Hot (One-Hot Encoding)**
   - **Descrição**: Cada valor categórico é representado por uma coluna binária (0 ou 1).
   - **Objetivo**: Evitar a introdução de ordens artificiais, mas pode aumentar significativamente o número de colunas.

3. **Codificação de Frequência (Frequency Encoding)**
   - **Descrição**: Cada valor categórico é substituído pela sua frequência de ocorrência, mantendo a importância relativa de cada categoria com base em sua frequência.
   - **Objetivo**: Preservar a distribuição dos dados categóricos, mantendo uma representação numérica simples.

4. **Codificação de Ordem (Ordinal Encoding)**
   - **Descrição**: Usada quando há uma ordem natural nos valores categóricos, cada valor é substituído por um número inteiro que respeita essa ordem.
   - **Objetivo**: Preservar a ordem natural dos dados categóricos.

#### Exemplo de Codificação One-Hot

**Cenário**: Suponha que temos um atributo categórico `Cor` com os valores: `Vermelho`, `Verde`, `Azul`.

**Antes da Codificação**:

| Cor     |
|---------|
| Vermelho|
| Verde   |
| Azul    |
| Verde   |
| Vermelho|

**Após a Codificação One-Hot**:

| Vermelho | Verde | Azul |
|----------|-------|------|
| 1        | 0     | 0    |
| 0        | 1     | 0    |
| 0        | 0     | 1    |
| 0        | 1     | 0    |
| 1        | 0     | 0    |

```python
import pandas as pd

# Dados de entrada
data = {'Cor': ['Vermelho', 'Verde', 'Azul', 'Verde', 'Vermelho']}
df = pd.DataFrame(data)

# Codificação One-Hot
df_encoded = pd.get_dummies(df, columns=['Cor'])

print(df_encoded)

# OBS: note, neste exemplo, a necessidade de limpeza dos dados pois, quaisquer mudança na descrição
#      de uma cor, geraria uma nova coluna (por exemplo, Vermelho e Vermelha)
```

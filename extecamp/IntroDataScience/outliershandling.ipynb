### Identificação e Remoção de Outliers
Este exemplo utiliza a biblioteca pandas para manipulação de dados e seaborn para visualização. 
O objetivo é identificar e remover outliers usando o método do intervalo interquartil (IQR).

### Passos:
1. Carregar um conjunto de dados.
2. Identificar outliers.
3. Visualizar outliers.
4. Remover outliers.

O exemplo utiliza um conjunto de dados fictício de preços de casas

##### Passo 1: Carregar o Conjunto de Dados
Criar um DataFrame fictício com alguns preços de casas, incluindo outliers.

```python
import pandas as pd

# Criando um DataFrame fictício
data = {'price': [30000, 120000, 130000, 125000, 145000, 500000, 135000, 140000, 1000000, 150000, 115000, 130000, 400000]}
df = pd.DataFrame(data)
print(df)
```

##### Passo 2: Identificar Outliers Usando IQR
Calcular o primeiro quartil (Q1), terceiro quartil (Q3), e o intervalo interquartil (IQR). 
Outliers são valores que estão abaixo de Q1 - 1.5 * IQR ou acima de Q3 + 1.5 * IQR.

```python
# Calculando Q1, Q3 e IQR
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

# Definindo os limites para detecção de outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")

# Identificando outliers
outliers = df[(df['price'] < lower_bound) | (df['price'] > upper_bound)]
print("Outliers encontrados:")
print(outliers)
```

##### Passo 3: Visualizar Outliers
Vamos usar seaborn para criar um boxplot e visualizar os outliers.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Personalizando a cor dos eixos
ax = plt.gca()  # Get the current axes instance
ax.spines['bottom'].set_color('white')  # Cor do eixo X
ax.spines['left'].set_color('white')  # Cor do eixo Y

# Personalizando a cor dos ticks e labels dos eixos (PARA FUNDO ESCURO)
ax.xaxis.label.set_color('white')  # Cor do label do eixo X
ax.yaxis.label.set_color('white')  # Cor do label do eixo Y
ax.tick_params(axis='x', colors='white')  # Cor dos ticks do eixo X
ax.tick_params(axis='y', colors='white')  # Cor dos ticks do eixo Y
ax.title.set_color('white')

# Criando um boxplot
sns.boxplot(y=df['price'])
plt.title('Boxplot de Preços de Casas')
plt.show()
```

##### Passo 4: Remover Outliers
Agora, vamos remover os outliers do DataFrame original.

```python
# Removendo outliers
df_cleaned = df[~((df['price'] < lower_bound) | (df['price'] > upper_bound))]

print("DataFrame sem outliers:")
print(df_cleaned)
```

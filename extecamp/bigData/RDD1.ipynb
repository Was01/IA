```python
import numpy as np
import pandas as pd

from pyspark.sql import SparkSession
```

```python
# Criação da sessão Spark

spark = (SparkSession.builder.master("local[*]").appName("PySpark Example")
.getOrCreate())

sc = spark.sparkContext

sc.setLogLevel("WARN") # (opcional)
```

```python
# Criando um RDD

rdd=sc.parallelize([1,2,3,4,5])
```

```python
rdd.collect()
```

```python
rdd2=sc.parallelize(['Washington','Fernandes','de','Barros'])
rdd2.collect()
```

```python
# Utilizando o MAP

rdd=sc.parallelize([1,2,3,4,5])
rdd2=rdd.map(lambda x: x**2)
rdd2.collect()
```

```python
# Utilizando a função filter

rdd=sc.parallelize(range(1,11))

rdd2=rdd.filter(lambda x:x%2==0)
rdd2.collect()
```

```python
# Utilizando a função FlapMap

rdd=sc.parallelize(['Washington Fernandes de Barros',"Professor de matemática"])
rdd.collect()
```

```python
rdd2=rdd.flatMap(lambda x:x.split())
rdd2.collect()
```

```python
# Utilizando Count, Max, Min, Sum, e Mean

rdd=sc.parallelize(range(1,51))
rdd.collect()
```

```python
rdd.count()
```

```python
rdd.min()
```

```python
rdd.max()
```

```python
rdd.sum()
```

```python
# Exercícios

# Crie um RDD a partir de uma lista de números de 1 a 10. Em seguida, mostre os elementos desse RDD.

rdd1=sc.parallelize(range(1,11))
rdd1.collect()
```

```python
# Utilize a função parallelize para criar um RDD a partir de uma lista de números de 1 a 100.

rdd2=sc.parallelize(range(1,101))
rdd2.collect()
```

```python
# Crie um RDD com números de 1 a 5 e use a função map() para criar um novo RDD com o quadrado de cada número.


rdd=sc.parallelize([1,2,3,4,5])
rdd3=rdd.map(lambda x:x**2)
rdd3.collect()
```

```python
# Crie um RDD com números de 1 a 20 e utilize a função filter() para criar um RDD contendo apenas números ímpares.

rdd=sc.parallelize(range(1,21))
rdd4=rdd.filter(lambda x:x%2==1)
rdd4.collect()
```

```python
# Crie um RDD com números de 1 a 10 e use a função reduce() para somar todos os números do RDD.

rdd=sc.parallelize(range(1,11))
rdd5=rdd.reduce(lambda x,y:x+y)
print(rdd5)
```

```python
# Crie um RDD a partir de uma lista de frases(strings). Use flatMap para dividir cada dividir cada frase em palavras e crir um RDD de palavras individuais.

rdd=sc.parallelize(['São Paulo','Rio de Janeiro','Paraná'])
rdd6=rdd.flatMap(lambda x: x.split())
rdd6.collect()
```

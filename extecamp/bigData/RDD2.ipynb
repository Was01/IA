```python

import numpy as np
import pandas as pd

from pyspark.sql import SparkSession
```

```python
# Criação da sessão Spark

spark = (SparkSession.builder.master("local[*]").appName("PySpark Example2")
.getOrCreate())

sc = spark.sparkContext

sc.setLogLevel("WARN") # (opcional)
```

```python
rdd = sc.parallelize([1,2,2,3,3,4,5,6])
rdd.collect()
```

```python
# Utilizando a função distinct

rdd.distinct().collect()
```

```python
# GroupBy() e GroupByKey()

rdd=sc.parallelize([('Raphael',300),('Raphael',200),('Bruno',150),('Regina',200)])

rdd.collect()
```

```python
rdd2=rdd.groupByKey()
rdd2=[(key,list(values)) for key,values in rdd2.collect()]
print(rdd2)
```

```python
rdd=sc.parallelize(['Raphael','Raphael','Bruno','Regina'])

rdd2=rdd.groupBy(lambda x: x[0])
rdd2=[(key,list(values)) for key,values in rdd2.collect()]
print(rdd2)
```

```python
 # keyBy()

rdd=sc.parallelize(['Raphael','Raphael','Bruno','Regina'])
rdd2=rdd.keyBy(lambda x:x[0])
rdd2.collect()
```

```python
rdd=sc.parallelize([('Raphael',300),('Raphael',200),('Bruno',150),('Regina',200)])

rdd2=rdd.reduceByKey(lambda x,y:x+y)
rdd2.collect()
```

## Exercícios

```python
# 1. Crie um RDD com os números [1,2,2,3,4,4,5]. Use distict() para remover os números duplicados.

rdd=sc.parallelize([1,2,2,3,4,4,5])

rdd.collect()
```

```python
rdd.distinct().collect()
```

```python
# 2. Dada uma lista de nomes ['Ana','Carlos','Ana','Maria','Carlos','José'] crie um RDD e aplique distict() para obter os nomes únicos.

rdd=sc.parallelize(['Ana','Carlos','Ana','Maria','Carlos','José'] )
rdd.collect()
```

```python
rdd.distinct().collect()
```

```python
# 3. Agrupando pares de chave valor pelo valor de chave

# Crie um RDD com pares chave-valor como [('a',1),('b',2),('a',3),('c',4),('b',5)]

rdd=sc.parallelize([('a',1),('b',2),('a',3),('c',4),('b',5)]
)

rdd.collect()
```

```python
rdd2=rdd.groupByKey()

rdd2=[(key,list(values)) for key,values in rdd2.collect()]
print (rdd2)
```

```python
# 4. Agrupando os nomes pela primeira letra.

# Dada uma lista de nomes ['Alice','Bob','Ana','Charlie','Daniel','Carl'], agrupe os nome pela primeira letra.

rdd=sc.parallelize(['Alice','Bob','Ana','Charlie','Daniel','Carl'])
rdd.collect()

```

```python
rdd2=rdd.groupBy(lambda x: x[0])
rdd2=[(key,list(values)) for key,values in rdd2.collect()]
print(rdd2)
```

```python
# 5. Agrupando uma lista de números, podemos agrupar os números em duas categorias par ou ímpar.

rdd=sc.parallelize(range(1,101))
rdd.collect()
```

```python
rdd2=rdd.groupBy(lambda x: 'par' if x%2==0 else 'ímpar')
rdd2=[(key,list(values)) for key,values in rdd2.collect()]
print(rdd2)
```

```python
# 6. Criando chave a partir da primeira letra da palavra

# Utilize KeyBy() para agrupar as palavras com base na primeira letra da palavra

rdd=sc.parallelize(['Ana','Carlos','Ana','Maria','Carlos','José'] )
rdd2=rdd.keyBy(lambda x:x[0] )
rdd2.collect()
```

```python
# 7. Dada uma lista de palavras, utilize keyBy() para agrupar as palavras com base no seu comprimento.

rdd=sc.parallelize(['Ana','Carlos','Ana','Maria','Carlos','José'] )
rdd2=rdd.keyBy(lambda x:len(x) )
rdd2.collect()
```

```python
# 8. Somr os valores de cada chave

# Dada uma lista de pares chave-valor, use reduceByKey para somar os valores de cada chave

rdd=sc.parallelize([('Raphael',300),('Raphael',200),('Bruno',150),('Regina',200)])

rdd2= rdd.reduceByKey(lambda x,y:x+y)

rdd2.collect()
```

```python
# 9. Encontrar o valor máximo por chave

# Dada uma lista de pares chave-valor, use reduceByKey() para encontrar o valor máximo para cada chave.

rdd=sc.parallelize([('Raphael',300),('Raphael',200),('Bruno',150),('Regina',200)])

rdd2= rdd.reduceByKey(lambda x,y:max(x,y))

rdd2.collect()
```

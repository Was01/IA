## 2. Descrição das variáveis

### 2.1 `data`

- **Descrição:** data de referência da observação.
- **Tipo:** data (armazenada como texto no CSV).
- **Periodicidade no arquivo:** diária.
- **Utilização:** identificação temporal das observações e ordenação da série histórica.

### 2.2 `Desemprego (%)`

- **Descrição:** taxa de desocupação da população economicamente ativa.
- **Tipo:** numérico (float).
- **Unidade:** percentual (%).
- **Periodicidade provável:** mensal ou trimestral móvel.
- **Interpretação econômica:** indica a proporção da força de trabalho que está desocupada e procura emprego.
- **Fonte provável:** IBGE — Pesquisa Nacional por Amostra de Domicílios Contínua (PNAD Contínua).

### 2.3 `Dívida bruta (% PIB)`

- **Descrição:** relação entre a dívida bruta do governo e o Produto Interno Bruto.
- **Tipo:** numérico (float).
- **Unidade:** percentual do PIB.
- **Periodicidade provável:** mensal.
- **Interpretação econômica:** representa o nível de endividamento do setor público em relação ao tamanho da economia.
- **Fonte provável:** Banco Central do Brasil.

### 2.4 `Produção industrial (índice)`

- **Descrição:** indicador da evolução da atividade industrial brasileira.
- **Tipo:** numérico (float).
- **Unidade:** número-índice.
- **Periodicidade provável:** mensal.
- **Interpretação econômica:** permite acompanhar a expansão ou contração da produção industrial ao longo do tempo.
- **Fonte provável:** IBGE — Pesquisa Industrial Mensal (PIM-PF).
- **Observação:** a base de referência e o tratamento sazonal do índice precisam ser confirmados.

### 2.5 `Vendas varejo (índice)`

- **Descrição:** indicador relacionado ao desempenho do comércio varejista.
- **Tipo:** numérico (float).
- **Unidade:** não confirmada.
- **Periodicidade provável:** mensal.
- **Interpretação econômica:** permite acompanhar mudanças na atividade do comércio varejista.
- **Fonte provável:** IBGE — Pesquisa Mensal de Comércio (PMC).
- **Observação:** os valores observados sugerem que a variável pode representar uma taxa de variação percentual, e não um índice em nível. Essa hipótese precisa ser confirmada antes da análise.

### 2.6 `IPCA (% a.m.)`

- **Descrição:** variação mensal do Índice Nacional de Preços ao Consumidor Amplo (IPCA).
- **Tipo:** numérico (float).
- **Unidade:** percentual ao mês (% a.m.).
- **Periodicidade provável:** mensal.
- **Interpretação econômica:** representa a inflação mensal medida pelo IPCA.
- **Fonte provável:** IBGE.

### 2.7 `PIB Trimestral (R$ mi)`

- **Descrição:** Produto Interno Bruto brasileiro em valores monetários.
- **Tipo:** numérico (float).
- **Unidade:** milhões de reais (R$ milhões).
- **Periodicidade provável:** trimestral.
- **Interpretação econômica:** representa o valor agregado dos bens e serviços finais produzidos na economia durante determinado período.
- **Fonte provável:** IBGE — Contas Nacionais Trimestrais.
- **Observação:** é necessário confirmar se os valores são nominais ou reais e se possuem ajuste sazonal.

### 2.8 `Selic (% a.a.)`

- **Descrição:** taxa de juros associada à política monetária brasileira.
- **Tipo:** numérico (float).
- **Unidade:** percentual ao ano (% a.a.).
- **Periodicidade provável:** diária ou associada às alterações da taxa.
- **Interpretação econômica:** representa uma referência fundamental para o custo do crédito, decisões de investimento e controle da inflação.
- **Fonte provável:** Banco Central do Brasil.
- **Observação:** é necessário confirmar se a série representa a meta da Selic ou a taxa Selic efetiva.

### 2.9 `Câmbio (R$/USD)`

- **Descrição:** cotação do dólar americano em relação ao real brasileiro.
- **Tipo:** numérico (float).
- **Unidade:** reais por dólar (R$/US$).
- **Periodicidade provável:** diária, em dias úteis.
- **Interpretação econômica:** indica quantos reais são necessários para adquirir um dólar americano.
- **Fonte provável:** Banco Central do Brasil.
- **Observação:** é necessário confirmar se a cotação utilizada corresponde à taxa de compra, venda ou PTAX.

```python

# 1. Apresentação

# Nome do sistema
print('Sistema para análise de glicose')


# Menu com as opções do sistema
print('\nMenu\n')
print('Opção 1: Dados da pessoa')
print('Opção 2: Classificação individual')
print('Opção 3: Análise da lista')
print('Opção 4: Relatório final')
```

```python
# 3. Dados e tipos

# Criação das variáveis nome, idade, glicose1, glicose2, e em_jejum. Para cada variável foi atribuído um valor arbitrário, respeitando os tipos de dados  solicitados nos requisitos do sistema (str,int, ou float, int ou float, bool)

nome='Jeniffer'
idade=39
glicose1=90
glicose2=130
em_jejum=False

#Impressão dos valores das variáveis e seus respectivos tipos

print('Nome:',nome,'| Tipo da variável: ',type(nome))
print('Idade:',idade,'| Tipo da variável: ',type(idade))
print('Glicose1: ' ,glicose1,'| Tipo da variável: ',type(glicose1))
print('Glicose2: ' ,glicose2,'| Tipo da variável: ',type(glicose2))
print('Em_jejum: ' ,em_jejum,'| Tipo da variável: ',type(em_jejum))
```

```python
# 4. Cálculos

# Cálculo e exibição da média das duas medições de glicose

media=(glicose1+glicose2)/2.0
print('Média:',media)

# Cálculo e exibição da diferença glicose2-glicose1

diferenca=glicose2-glicose1
print('Diferença:',diferenca)

# Import da biblioteca math
import math

# Cálculo e impressão do resultado de math.sqrt(glicose) como teste de uso da biblioteca matemática

raiz1=math.sqrt(glicose1)
raiz2=math.sqrt(glicose2)

print('Raiz1:',raiz1)
print('Raiz2:',raiz2)

```

```python
# 5. Comparações

# Impressão dos resultados booleanos das comparações solicitadas nos requisitos do sistema

print("Média <= 70:",media<=70)
print("Média >= 100:",media>=100)
print("Idade >= 18:",idade>=18)
print("Glicose1==Glicose2:",glicose1==glicose2)
```

```python
# 6. Classificação da média de acordo com os requisitos do sistema

if media<=70:
  print('Pessoa com hipoglicemia')
elif media>=100:
  print('Pessoa com hiperglicemia')
else:
  print('Pessoa sem hipo ou hiperglicemia')
```

```python
# 7. Verificação da maioridade

if idade>=18:
  print('Atingiu a maioridade')
else:
  print('Pessoa menor de idade')
```

```python
# 8. Operadores lógicos

# Variável alerta recebe True quando a média está fora da faixa intermediária (media<=70 ou media>=100) e a pessoa está em jejum

alerta = (media<=70 or media>=100) and em_jejum

print('Alerta:',alerta)
print('Sem alerta:',not alerta)
```

```python
# 9. Análise de lista

glicemias=[65,72,89,100,118,70,99]

# Percorre a lista com for e imprime cada valor com sua respectiva classificação

for glicemia in glicemias:
  if glicemia<=70:
    print(glicemia,'- Pessoa com hipoglicemia')
  elif glicemia>=100:
    print(glicemia,'- Pessoa com hiperglicemia')
  else:
    print(glicemia,'- Pessoa sem hipo ou hiperglicemia')

```

```python
# 10. Contadores e relatório final

glicemias=[65,72,89,100,118,70,99]

quantidade =0
soma=0

# Percorre a lista com for e imprime cada valor com sua respectiva classificação

for glicemia in glicemias:
  # soma acumula os valores da lista; quantidade conta quantos foram lidos
  quantidade+=1
  soma+=glicemia
  if glicemia<=70:
    print(glicemia,'- Pessoa com hipoglicemia')
  elif glicemia>=100:
    print(glicemia,'- Pessoa com hiperglicemia')
  else:
    print(glicemia,'- Pessoa sem hipo ou hiperglicemia')

# Relatório final com a soma total e a média absoluta

print('\nRelatório final')
print('Soma total:',soma)
print('Média absoluta da soma:',soma/quantidade)
```

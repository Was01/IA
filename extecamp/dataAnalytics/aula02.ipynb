```python
# Estrutura de repetição: laço for
# Útil para percorrer uma lista

for i in [10,20,30]:
  print(f'Valor de {i}')


```

```python
lista=[18,17,60]
for idade in lista:
  print('\nIdade:',idade)
  if(idade>=60):
    print('Pessoa idosa.')
  elif idade<18:
    print('Pessoa menor de idade.')
  else:
    print('Pessoa com idade maior ou igual a 18 anos e menor do que 60 anos.')

```

```python
# Manipulação de listas e índices: comando enumerate

valores=[10,20,30,40]
for indice, valor in enumerate(valores):
  #print(indice,valor)
  if indice==2:
    print('Elemento da posição 2:',valor)
```

```python
# Dicionários e registros

registros=[
    {'Id':0,'idade':20,'altura':1.70},
    {'Id':1,'idade':25,'altura':1.6},
    {'Id':2,'idade':30,'altura':1.55}]

print(registros) #exibe todos os registros
```

```python
for pessoa in registros:
  for chave, valor in pessoa.items():
    print(f'{chave}:{valor}')

```

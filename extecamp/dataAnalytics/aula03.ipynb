```python
# importação das bibliotecas

import numpy as np
import pandas as pd
from google.colab import files
files.upload()
```

```python
# Carregando banco de dados

arquivo='/content/pima_diabetes.csv'

df=pd.read_csv(arquivo)
df.head()
```

```python
# Visualizar as colunas do datset

print(df.columns)
print(df.shape)
```

```python
print(df.info)
print(df.dtypes)
```

```python
df.describe()
```

```python
# Relizar as análises com a variável glicose do dataset

dados_glicose=df['Glucose']
dados_glicose
```

```python
dados_glicose.describe()
```

```python
# Média mediana e desvio padrão variável glicose

media=dados_glicose.mean()
mediana=dados_glicose.median()
desvio_padrao=dados_glicose.std()
print('Média: ',media)
print('Mediana: ',mediana)
print('Desvio padrão: ',desvio_padrao)
```

```python
# Média mediana e desvio padrão variável BMI

indice_massa_corporal=df['BMI']

media=indice_massa_corporal.mean()
mediana=indice_massa_corporal.median()
desvio_padrao=indice_massa_corporal.std()
print('Média: ',media)
print('Mediana: ',mediana)
print('Desvio padrão: ',desvio_padrao)
```

```python
# Verificar quais valores estão com zero

zeros=dados_glicose[dados_glicose==0]

print(zeros)
```

```python
# Verificar quais valores não foram preenchidos na coluna glicose

nulos=dados_glicose.isnull().sum()
print(nulos)
```

```python
# Carregar os dados de glicose e realizar a filtragem para excluir
# campos vazios ou com valores incoerentes

dados_glicose_filtrados = dados_glicose.replace(0, np.nan).dropna()
```

```python
print(dados_glicose_filtrados.mean())
print(dados_glicose_filtrados.std())
```

```python
# Geração de gráficos para coluna glicose
import pandas as pd
import matplotlib.pyplot as plt

plt.hist(dados_glicose_filtrados,bins=20)
plt.axvline(dados_glicose_filtrados.mean(),color='red',label='média')
plt.show()
```

```python
# Grafico com média +-1
plt.bar(0,dados_glicose_filtrados.mean(),color='blue')
plt.errorbar(0,dados_glicose_filtrados.mean(),yerr=dados_glicose_filtrados.std(),color='red',capsize=8)
plt.xticks([0],['Glucose'])
plt.show()
```

```python
# Gerar um gráfico com média e desvio padrão para variável BMI

bmi=pd.read_csv('/content/pima_diabetes.csv')['BMI']

bmi_filtrados=bmi.replace(0,pd.NA).dropna()

plt.bar(0,bmi_filtrados.mean(),color='blue')
plt.errorbar(0,bmi_filtrados.mean(),yerr=bmi_filtrados.std(),color='black',capsize=8)
plt.xticks([0],['BMI'])
plt.show()
```

```python
# BoxPlot


plt.figure(figsize=(10,8))
plt.boxplot(dados_glicose_filtrados,vert=False)
plt.title(f'Boxplot')
plt.xlabel('Glicose')
plt.show()
```

```python
dados_glicose_filtrados.describe()
```

```python
# Cálculo de skewness: Medidas de assimetria

skew=dados_glicose_filtrados.skew()
print(skew)
```

```python
# Gerar o gráfico

plt.hist(dados_glicose_filtrados,bins=20,edgecolor='black')
plt.axvline(dados_glicose_filtrados.mean(),label='Média',color='red')
plt.axvline(dados_glicose_filtrados.mean(),label='Mediana',color='green')
plt.show()
```
